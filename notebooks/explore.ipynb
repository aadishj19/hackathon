# Explore the challenge data

Every file in `DATA_DIR` (see `.env`) is loaded as a table. Query it with SQL (`con.sql`) or pull it into pandas (`.df()`). When a table is clean, export it for Power BI with the last cell.

```python
from hack import data

con = data.connect()
data.tables(con)
```

```python
print(data.describe(con))
```

```python
df = con.sql("""
    SELECT c.segment, t.category, ROUND(SUM(-t.amount_eur)) AS spend_eur
    FROM transactions t JOIN customers c USING (customer_id)
    GROUP BY ALL ORDER BY spend_eur DESC
""").df()
df.head(10)
```

```python
# Save a cleaned result as its own table, then export everything for Power BI (exports/*.csv).
# DuckDB reads the pandas variable `df` directly by name.
con.execute("CREATE OR REPLACE TABLE spend_by_segment AS SELECT * FROM df")
data.export_for_powerbi(con)
```
